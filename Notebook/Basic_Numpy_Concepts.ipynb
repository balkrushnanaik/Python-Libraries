# 4. Array Indexing and Slicing

1. Array indexing is the same as accessing an array element.
2. You can access an array element by referring to its index number.
3. The indexes in NumPy arrays start with 0, meaning that the first element has index 0, and the second has index 1 etc.

```python
arr_in = np.array([1,2,3,4,5,6])
print(arr_in)

#Sum of array indexing
sum_in = arr_in[0] + arr_in[1]
print(sum_in)
```

4.2 Array Slicing

Slicing in python means taking elements from one given index to another given index.

We pass slice instead of index like this: [start:end].

We can also define the step, like this: [start:end:step].

If we don't pass start its considered 0

If we don't pass end its considered length of array in that dimension

If we don't pass step its considered 1

```python
arr_sl = np.array([1,2,3,4,5,6])
print(arr_sl)

print(arr_sl[1::2])

# (Start: end: step)
```

```python
import numpy as np
arr_sl = np.array([1,2,3,4,5,6])
print(arr_sl)
print(arr_sl[::-1])
```

```python
arr2d = np.random.randint(40,100,(5,2)) #(rows,columns)
print(arr2d)
```

```python
print(arr2d[1]) # specific element - 1D
print(arr2d[1:]) # slicing part - 2D
```

# 5 Array Reshaping and Flattening

#### Reshape Array: 
- changing its dimensions (e.g., rows and columns) without altering the actual data
#### Flatten Array:
- transforming a multi-dimensional array into a single-dimensional array

```python
arr1 = np.array([1,2,3,4,5,6])
print(type(arr1))
print(arr1.ndim)
print(arr1)
```

```python
reshaped = arr1.reshape((6,1))      # shape = (6,1) - 6 rows and 1 column
print(reshaped)
```

```python
reshaped2 = arr1.reshape((2,3))     # re-shaped array from (1,6) to (2,3)
print(reshaped2)
```

# 6 Array Stacking and Splitting

#### Array Stacking: 
- combining multiple arrays along a new axis, resulting in an array with a higher dimension than the input arrays
#### Array Splitting:
- Dividing a single array into multiple smaller arrays along a specified axis.

```python
# stacking
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b))) # vertical stacking - row wise          -> output array is 2D
print(np.hstack((a,b))) # horizontal stacking - column wise  
print(np.stack((a,b)))
```

```python
# splitting
c = np.array([[1,2,3], [4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3) # horizontal split -> input and output array are same dimention

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # vertical split -> input and output array are same dimention

for s in vsplit:
    print(s)
```

## 11 Broadcasting

##### Broadcasting is the process by which NumPy automatically expands smaller arrays or scalar values to match the shape of larger arrays, allowing arithmetic operations without copying data.

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([10,20,30])
print(a+b)
```
